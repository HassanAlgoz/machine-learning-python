## 2. Auto MPG Dataset

The Auto MPG Dataset is a classic benchmark for regression analysis in machine learning. It originally appeared in the 1983 American Statistical Association (ASA) Exposition and was later donated to the UCI Machine Learning Repository by Ross Quinlan in 1993.

The data consists of technical specifications for various car models from the late 1970s and early 1980s, primarily used to predict fuel efficiency (MPG).

- Features: `5` numerical, `3` categorical
- Target: `mpg` (miles per gallon)
- Size: `398` samples
- Source: [Auto MPG Dataset](https://archive.ics.uci.edu/dataset/9/auto+mpg)

```python
pd.read_csv("../datasets/auto-mpg.csv").head()
```

## 3. California Housing Prices Dataset

The California Housing Prices Dataset is a staple in regression-based machine learning, derived from the 1990 U.S. Census data. It is the modern, ethically safer alternative to the deprecated Boston Housing dataset.

According to the official Scikit-learn documentation and the original paper by Pace and Barry (1997), each record represents one "census block group"—the smallest geographical unit for which the Bureau publishes sample data.

- Features: `8` numerical, `1` categorical (`ocean_proximity`)
- Target: `median_house_value` (in thousands of dollars).
- Size: `20,640` samples
- Source: [California Housing Dataset](https://media.geeksforgeeks.org/wp-content/uploads/20240522145850/housing%5B1%5D.csv)
