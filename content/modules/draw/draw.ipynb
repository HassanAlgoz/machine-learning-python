```python
import matplotlib.pyplot as plt
import numpy as np

# إعداد البيانات والنطاق المطلوبه
x = np.linspace(-2, 10, 100)

# تعريف المعادلات (y = mx + b) مع عناوين واضحة بالإنجليزية والمعادلة الرياضية
cases = [
    (2 * x + 1, 'Proportional (m=2)\n$y = 2x + 1$', 'green'),
    (0.5 * x + 3, 'Proportional (m=0.5)\n$y = 0.5x + 3$', 'lightgreen'),
    (-1 * x + 5, 'Inverse (m=-1)\n$y = -x + 5$', 'red'),
    (-3 * x + 10, 'Inverse (m=-3)\n$y = -3x + 10$', 'darkred'),
    (0 * x + 4, 'Constant (m=0)\n$y = 4$', 'blue')
]

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

for i, (y, title, color) in enumerate(cases):
    row, col = divmod(i, 3)
    ax = axes[row, col]
    ax.plot(x, y, color=color, linewidth=2)
    
    # تحسين وضوح العنوان وحجمه
    ax.set_title(title, fontsize=14)
    
    # إضافة تسميات واضحة للمحاور بخط أكبر
    ax.set_xlabel('x (Independent)', fontsize=12)
    ax.set_ylabel('y (Dependent)', fontsize=12)
    
    # تثبيت النطاق
    ax.set_xlim([-2, 10])
    ax.set_ylim([-2, 10])
    
    # تحسين حجم الأرقام على المحاور
    ax.tick_params(axis='both', which='major', labelsize=10)

axes[1, 2].axis('off')
plt.tight_layout()
plt.show()
```

import numpy as np
import matplotlib.pyplot as plt

# Set style and reproducibility
plt.style.use('ggplot')
np.random.seed(42)
x_smooth = np.linspace(0, 10, 200)
x_points = np.linspace(0.5, 9.5, 30)

def plot_and_save_relationship(x_s, y_s, x_p, y_p, title, filename):
    plt.figure(figsize=(10, 6))
    plt.scatter(x_p, y_p, color='black', alpha=0.7, label='Noisy Data')
    plt.plot(x_s, y_s, color='blue', linewidth=3, label='Underlying Function')
    plt.title(title, fontsize=16)
    plt.xlabel('X Variable', fontsize=12)
    plt.ylabel('Y Variable', fontsize=12)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename)
    plt.show()
    plt.close()

# 1. Quadratic Relationship
y_quad_smooth = -(x_smooth - 5)**2 + 35
y_quad_points = -(x_points - 5)**2 + 35 + np.random.normal(0, 2, len(x_points))
plot_and_save_relationship(x_smooth, y_quad_smooth, x_points, y_quad_points,
                           'Quadratic Relationship\nWorking Hours vs. Productivity',
                           'quadratic_relationship.png')

# 2. Cubic Relationship
y_cubic_smooth = 0.5 * (x_smooth - 5)**3 + 50
y_cubic_points = 0.5 * (x_points - 5)**3 + 50 + np.random.normal(0, 5, len(x_points))
plot_and_save_relationship(x_smooth, y_cubic_smooth, x_points, y_cubic_points,
                           'Cubic Relationship\nDistinct Curves',
                           'cubic_relationship.png')

# 3. Exponential Relationship
y_exp_smooth = 2 * np.exp(0.4 * x_smooth)
y_exp_points = 2 * np.exp(0.4 * x_points) + np.random.normal(0, 4, len(x_points))
plot_and_save_relationship(x_smooth, y_exp_smooth, x_points, y_exp_points,
                           'Exponential Relationship\nRapid Growth',
                           'exponential_relationship.png')

# 4. Logarithmic Relationship
y_log_smooth = 20 * np.log(x_smooth + 1)
y_log_points = 20 * np.log(x_points + 1) + np.random.normal(0, 3, len(x_points))
plot_and_save_relationship(x_smooth, y_log_smooth, x_points, y_log_points,
                           'Logarithmic Relationship\nLearning Curve',
                           'logarithmic_relationship.png')

# 5. Cosine Relationship
y_cos_smooth = 15 * np.cos(x_smooth) + 20
y_cos_points = 15 * np.cos(x_points) + 20 + np.random.normal(0, 2, len(x_points))
plot_and_save_relationship(x_smooth, y_cos_smooth, x_points, y_cos_points,
                           'Cosine Relationship\nWave Pattern',
                           'cosine_relationship.png')
